```python
import re

# Function to extract the recipe name, secondary name (in parentheses), ingredients, method of preparation, and health benefits
def extract_recipes_with_health_benefits(text):
    recipes = []
    
    # Split the text by known separator patterns for each recipe
    recipe_blocks = re.split(r'TRADITIONAL FOOD RECIPES from AYUSH SYSTEMS of MEDICINE AYUSHMAN BHARAT', text)
    
    for block in recipe_blocks:
        recipe = {}
        
        # Extract Recipe Name: Look for a name followed by an English name in parentheses (like "KHALAM (Medicated Buttermilk)")
        name_match = re.search(r'([A-Z\s]+)\s+\(([^)]+)\)', block)
        if name_match:
            recipe['recipe_name'] = name_match.group(1).strip()  # Primary name (e.g., "KHALAM")
            recipe['secondary_name'] = name_match.group(2).strip()  # Secondary name (e.g., "Medicated Buttermilk")
        else:
            # If no secondary name, just capture the primary name
            name_match = re.search(r'([A-Z\s]+)', block)
            if name_match:
                recipe['recipe_name'] = name_match.group(1).strip()

        # Extract Ingredients and Quantity
        ingredients_section = re.search(r'Ingredients\s+Quantity(.*?)Method of preparation', block, re.DOTALL)
        if ingredients_section:
            ingredients_raw = ingredients_section.group(1).strip()
            ingredients_list = re.findall(r'([\w\s\(\)-]+)\s+-\s+([\w\s\/]+)', ingredients_raw)
            recipe['ingredients'] = [{'ingredient': item[0].strip(), 'quantity': item[1].strip()} for item in ingredients_list]
        
        # Extract Method of Preparation
        method_section = re.search(r'Method of preparation(.*?)Health beneﬁts', block, re.DOTALL)
        if method_section:
            recipe['method'] = method_section.group(1).strip()
        else:
            recipe['method'] = None  # Set to None if method isn't found

        # Extract Health Benefits
        health_benefits_section = re.search(r'Health beneﬁts(.*?)\n', block, re.DOTALL)
        if health_benefits_section:
            recipe['health_benefits'] = health_benefits_section.group(1).strip()
        else:
            recipe['health_benefits'] = None  # Set to None if health benefits are not found

        # Append the recipe to the list if a valid name and ingredients are found
        if recipe.get('recipe_name') and recipe.get('ingredients'):
            recipes.append(recipe)
    
    return recipes

# Read the content from the text file (replace 'path_to_text_file.txt' with your actual file path)
with open('/Users/shravyadsouza/Desktop/cleaned_output-fitz.txt', 'r', encoding='utf-8') as file:
    text = file.read()

# Extract recipes from the text
extracted_recipes = extract_recipes_with_health_benefits(text)

# Display extracted recipes
for recipe in extracted_recipes:
    print(f"Recipe Name: {recipe.get('recipe_name')}")
    print(f"Secondary Name: {recipe.get('secondary_name', 'N/A')}")
    print("Ingredients:")
    for ingredient in recipe['ingredients']:
        print(f" - {ingredient['ingredient']}: {ingredient['quantity']}")
    
    # Print method if available
    if recipe.get('method'):
        print(f"Method of Preparation: {recipe['method']}")
    else:
        print("Method of Preparation: Not available")
    
    # Print health benefits if available
    if recipe.get('health_benefits'):
        print(f"Health Benefits: {recipe['health_benefits']}")
    else:
        print("Health Benefits: Not available")

    print("\n" + "-"*40 + "\n")
```

```python
import csv

# Function to save extracted recipes to a CSV file
def save_recipes_to_csv(recipes, file_path):
    # Define the CSV file columns
    columns = ['Recipe Name', 'Secondary Name', 'Ingredients', 'Method of Preparation', 'Health Benefits']

    # Open a CSV file for writing
    with open(file_path, mode='w', newline='', encoding='utf-8') as file:
        writer = csv.DictWriter(file, fieldnames=columns)
        writer.writeheader()  # Write the header (column names)

        # Write each recipe as a row in the CSV
        for recipe in recipes:
            # Join ingredients into a single string
            ingredients_str = '; '.join([f"{ing['ingredient']}: {ing['quantity']}" for ing in recipe['ingredients']])
            
            # Write the recipe row
            writer.writerow({
                'Recipe Name': recipe.get('recipe_name'),
                'Secondary Name': recipe.get('secondary_name', 'N/A'),
                'Ingredients': ingredients_str,
                'Method of Preparation': recipe.get('method', 'Method of Preparation not available'),
                'Health Benefits': recipe.get('health_benefits', 'Health Benefits not available')
            })

    print(f"Recipes successfully saved to {file_path}")
```

```python
# Read the content from the text file (replace 'path_to_text_file.txt' with your actual file path)
with open('/Users/shravyadsouza/Desktop/cleaned_output-fitz.txt', 'r', encoding='utf-8') as file:
    text = file.read()

# Extract recipes from the text
extracted_recipes = extract_recipes_with_health_benefits(text)

# Save the extracted recipes to a CSV file
csv_file_path = '/Users/shravyadsouza/Desktop/extracted_recipes.csv'
save_recipes_to_csv(extracted_recipes, csv_file_path)
```
